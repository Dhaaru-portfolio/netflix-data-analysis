```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv('netflix_titles.csv')
print("Rows:", df.shape[0])
df.head()

# Q1: Movies vs TV Shows
plt.figure(figsize=(6,4))
df['type'].value_counts().plot(kind='bar', color=['#E50914','black'])
plt.title('Netflix - Movies vs TV Shows', fontsize=14, weight='bold')
plt.xticks(rotation=0)
plt.show()

# Q2: Top 10 Countries
plt.figure(figsize=(8,4))
df['country'].dropna().str.split(', ').explode().value_counts().head(10).plot(kind='barh', color='red')
plt.title('Top 10 Countries Producing Content')
plt.show()

# Q3: Content added by Year
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df['year_added'] = df['date_added'].dt.year
plt.figure(figsize=(8,4))
df['year_added'].value_counts().sort_index().plot(marker='o', color='#E50914')
plt.title('Content Added Trend (2008-2021)')
plt.show()

# Q4: Top Genres
plt.figure(figsize=(8,4))
df['listed_in'].str.split(', ').explode().value_counts().head(10).plot(kind='bar', color='black')
plt.title('Top 10 Genres on Netflix')
plt.xticks(rotation=45)
plt.show()

print("PROJECT 2 DONE DHEE! 🎉")
```
